```python
import matplotlib.pyplot as plt

def my_function(x):
    return x**2  # Example function: f(x) = x^2

x_values = [i/10 for i in range(-5, 6)]  # Generate x values from -0.5 to 5 in increments of 0.1
y_values = [my_function(x) for x in x_values]  # Compute y values based on the function

plt.plot(x_values, y_values)  # Plotting the function
plt.xlabel('x')             # Labeling the x-axis
plt.ylabel('f(x)')          # Labeling the y-axis
plt.title('Plot of f(x) = x^2')  # Title of the plot
plt.show()                  # Displaying the plot
```
